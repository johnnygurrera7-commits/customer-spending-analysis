# Business Intelligence Analysis: Customer Spending Behavior

## Introduction

This project analyzes customer spending behavior using transactional data.
The goal is to identify patterns in revenue based on time, customer demographics, and behavioral factors.

## Data Overview

```python
import pandas as pd

df = pd.read_csv("../data/sales_data.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

## Exploratory Analysis

This analysis compares average spending across customer genders.

```python
df.groupby("sex")["total_bill"].mean()
```

This analysis compares average spending across different times.

```python
df.groupby("time")["total_bill"].mean()
```

This analysis compares average spending across different days.

```python
df.groupby("day")["total_bill"].mean()
```

This analysis compares average spending across customers who smoke and those who don't. 

```python
df.groupby("smoker")["total_bill"].mean()
```

## Visualizations

## Revenue by Day

This chart shows average spending across different days of the week.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.barplot(x="day", y="total_bill", data=df, errorbar=None)

plt.title("Average Revenue by Day")
plt.xlabel("Day of Week")
plt.ylabel("Average Total Bill")

plt.show()
```

## Revenue by Day and Gender

This chart shows average spending across different days of the week across customer gender groups.

```python
plt.figure(figsize=(8,5))

sns.barplot(x="day", y="total_bill", hue="sex", data=df, errorbar=None)

plt.title("Average Revenue by Day and Gender")
plt.xlabel("Day of Week")
plt.ylabel("Average Total Bill")

plt.show()
```

## Revenue by Time of Day

This chart shows average spending across different times of the day.

```python
plt.figure(figsize=(6,4))

sns.barplot(x="time", y="total_bill", data=df, errorbar=None)

plt.title("Average Revenue by Time of Day")
plt.xlabel("Time")
plt.ylabel("Average Total Bill")

plt.show()
```

## Key Insights

1. **Weekend Revenue Trends**  
Average revenue peaks on weekends, particularly Sunday, suggesting increased customer spending during leisure periods at the end of the workweek.

2. **Customer Spending by Gender**  
Male customers exhibit slightly higher average spending compared to female customers across most days, although the difference is relatively moderate.

3. **Time-Based Spending Behavior**  
Dinner service consistently generates higher average revenue than lunch, indicating that evening dining is associated with higher-value transactions.

## Conclusion

This analysis demonstrates that customer spending behavior is influenced by temporal and demographic factors. 
Higher revenue is observed during weekends and dinner periods, while demographic differences exist but are relatively moderate. 
These insights can help guide business decisions related to staffing, promotions, and service optimization.
